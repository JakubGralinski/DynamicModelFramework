```python
import os
import json

input_path = os.getenv('INPUT_PATH', 'src/main/resources/intermediate.json')
output_path = os.getenv('OUTPUT_PATH', 'src/main/resources/results.json')

with open(input_path, 'r') as f:
    data = json.load(f)

fields_to_reset = ["PKB", "EKS", "IMP", "KI", "KS", "INW", "twKI", "twKS", "twINW", "twEKS", "twIMP", "NET_EXPORTS"]
fields_to_reset = [field for field in fields_to_reset if field in data]
for field in fields_to_reset:
    data[field] = [0] * len(data[field])

ZDEKS = [eks / pkb if pkb != 0 else 0 for eks, pkb in zip(data.get("EKS", []), data.get("PKB", []))]
data["ZDEKS"] = ZDEKS

with open(output_path, 'w') as f:
    json.dump(data, f, indent=4)
```
